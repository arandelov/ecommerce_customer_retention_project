# Customer Segmentation and Its Link to Retention Modeling

## 1. Objective

The goal of customer segmentation is to group customers based on behavioral and transactional patterns derived from their purchase history. These segments are later used to better interpret and validate the results of the retention prediction model.

---

## 2. Feature Space for Segmentation

Each customer is represented by a feature vector:

\[
\mathbf{x}_i =
\left[
\text{Purchase Frequency},
\text{Recency},
\text{Lifetime Value},
\text{Order Variability},
\text{Engagement Metrics},
\text{Demographics}
\right]
\]

Where:

- Purchase Frequency:
\[
\text{Purchase Rate} = \frac{\text{Total Purchases}}{\text{Customer Lifetime}}
\]

- Recency:
\[
\text{Recency Ratio} = \frac{\text{Days Since Last Purchase}}{\text{Customer Lifetime}}
\]

- Engagement:
\[
\text{Engagement Rate} = \frac{\text{Pages Viewed}}{\text{Session Duration}}
\]

---

## 3. Clustering Methodology

We apply unsupervised learning to discover natural customer groups:

\[
\text{Cluster}(x_i) = \arg \min_k \| x_i - \mu_k \|^2
\]

where $\mu_k$ represents the centroid of cluster $k$.

Typical algorithms:

- K-Means clustering
- Hierarchical clustering (for interpretability validation)

---

## 4. Business Interpretation of Clusters

After clustering, each segment is assigned a business label:

| Cluster | Behavioral Profile | Business Label |
|----------|-------------------|----------------|
| High frequency, high spend, low recency | Frequent repeat buyers | **Loyal Customers** |
| High spend, low frequency | Large but infrequent purchases | **High-Value Occasional Buyers** |
| Low spend, high recency | Inactive users | **At-Risk Customers** |
| Moderate activity across all metrics | Stable engagement | **Regular Customers** |

---

## 5. Connection to Retention Model

The segmentation output is not used as a feature in training but serves as an **external validation layer**:

\[
P(\text{Retention}_i = 1 \mid \mathbf{x}_i)
\]

We evaluate whether predicted retention aligns with observed cluster behavior:

- Loyal clusters → high predicted retention probability
- At-risk clusters → low predicted retention probability

This ensures interpretability and consistency between:

- supervised learning (retention prediction)
- unsupervised learning (customer segmentation)

---

## 6. Key Insight

Segmentation provides **behavioral structure**, while the retention model provides **predictive power**.

Together they form a complete customer intelligence system:

\[
\text{Segmentation} + \text{Prediction} = \text{Actionable CRM Strategy}
\]
