## 🧠 Feature Engineering Strategy for Retention Modeling (Leakage-Safe Design)

### 🎯 Goal
The objective is to predict whether a customer will become a **high-frequency buyer (≥ 3 purchases within 6 months)** using only information available at or shortly after the **first purchase**.  
This ensures the model reflects a real-world prediction scenario and avoids data leakage from future behavior.

---

## ❗ Key Principle: Avoid Data Leakage

Any feature that is computed using the **full 6-month observation window** must NOT be used as an input feature, because it already contains future information that the model is trying to predict.

### 🚫 Examples of leakage-prone features (DO NOT USE)
These features summarize behavior over the entire 6-month period:

- `Total_Spend`
- `Total_Quantity`
- `Avg_Discount_Rate`
- `Avg_Session_Duration_Minutes`
- `Avg_Pages_Viewed`

These variables directly or indirectly encode future behavior and would artificially inflate model performance.

---

## ✅ Recommended Feature Set (Leakage-Safe)

### 🟢 1. First Purchase / Acquisition Features
These are available at time 0 and represent initial customer intent:

- `Age`
- `Gender`
- `City`
- `Device_Type`
- `Payment_Method`
- `Product_Category` (first purchase)
- `Unit_Price` or `Total_Amount` (first order value)
- `Quantity` (first order quantity)
- `Discount_Rate` (first purchase discount usage)

👉 These features capture the customer's **initial buying behavior and profile**

---

### 🟢 2. Early Behavioral Signals (if available in short window)

Instead of full-window aggregates, use early-time behavior:

- Session duration from first session or first visit
- Pages viewed in first session
- Delivery time of first order
- Customer rating after first purchase (if immediate)

---

### 🟡 3. Engineered Early Lifecycle Features (Highly Recommended)

If transaction timestamps allow, create stronger predictive signals such as:

- `First_7_Days_Orders`
- `First_14_Days_Spend`
- `Days_To_Second_Order`
- `Early_Repeat_Indicator` (binary: second purchase within X days)

👉 These features are among the **strongest predictors of retention in real-world systems**

---

## 🧩 Correct Modeling Framing

This project should be interpreted as:

> ❌ "Explaining retention using full customer history"

---

## 🧩 Customer Segmentation (Unsupervised Learning)

### Objective
Before building predictive models, we perform customer segmentation to identify distinct behavioral groups based on purchasing behavior, value contribution, and lifecycle patterns.

This allows us to better understand *who the customers are*, complementing the retention prediction task.

---

### 🎯 Why segmentation?

Segmentation helps to:
- Identify homogeneous customer groups
- Understand differences in revenue and behavior patterns
- Support targeted marketing strategies
- Complement supervised retention modeling with behavioral insights

---

### 📊 Feature Selection for Segmentation

We use behavioral, monetary, and lifecycle variables:

\[
X = \{ \text{Total\_Num\_Orders}, \text{Revenue}, \text{AOV}, \text{Recency\_Days}, \text{Customer\_Lifetime\_Days}, \text{Avg\_Discount\_Rate} \}
\]

Optionally, RFM-based features can also be included:

\[
\{ R\_score, F\_score, M\_score, RFM\_Score \}
\]

---

### ⚙️ Data Preprocessing

Since clustering is distance-based, feature scaling is required:

\[
X_{scaled} = \frac{X - \mu}{\sigma}
\]

We apply standardization to ensure all variables contribute equally to distance calculations.

---

### 🧠 Clustering Method: K-Means

We apply K-Means clustering:

\[
\min \sum_{i=1}^{k} \sum_{x \in C_i} \|x - \mu_i\|^2
\]

where:
- \( k \) = number of clusters
- \( C_i \) = cluster \( i \)
- \( \mu_i \) = centroid of cluster \( i \)

---

### 📌 Choosing Optimal Number of Clusters

We use the **Elbow Method**, selecting \( k \) such that:

\[
\text{Inertia}(k) \text{ shows diminishing returns after optimal } k
\]

---

### 🧾 Cluster Interpretation

After fitting the model, we analyze cluster profiles by computing:

- Average revenue per segment
- Average number of orders
- Recency and lifetime behavior
- Segment size distribution

This allows us to assign meaningful business labels such as:

- High-value loyal customers
- Occasional high spenders
- Low-value inactive users
- Growth potential customers

---

### 📈 Business Validation

We validate segmentation by linking clusters to retention:

\[
\text{Retention Rate}_{segment} = \mathbb{E}[Retention \mid Segment]
\]

This helps identify:
- Which customer types are most loyal
- Which segments require activation strategies
- Where retention efforts should be focused

---

### 🎯 Expected Outcome

We expect to find:
- A small group of high-value loyal customers
- A large mid-tier segment driving most revenue
- A low-value inactive segment with high churn risk

These segments will directly support targeted retention strategies and improve model interpretability.

## 🎯 Modeling Strategy for Customer Retention

### Objective
The goal of this stage is to build predictive models that classify whether a customer will be retained (≥3 purchases within 6 months), using behavioral, monetary, and lifecycle features derived during EDA.

---

### Key Insights from EDA

- Retention is primarily **frequency-driven** (Total Orders, F_score, RFM Score)
- **Revenue correlates with retention**, but mainly through purchase frequency
- **AOV has little impact** on retention
- **Engagement metrics** (session duration, pages viewed) show minimal predictive power
- **Recency and lifetime** are not strong standalone predictors
- Customer base is **balanced**, not dominated by a small VIP segment

---

### Feature Selection Strategy

#### Include:
- Behavioral features:
  - `Total_Num_Orders`
  - `F_score`, `RFM_Score`
- Monetary features:
  - `Revenue`, `M_score`
- Discount behavior:
  - `Avg_Discount_Rate`
- Categorical features:
  - `City`, `Most_Used_Device`, `Most_Used_Payment`
- Optional:
  - `Recency_Days`, `Customer_Lifetime_Days` (test impact)

#### Exclude or Deprioritize:
- `Avg_Session_Duration_Minutes`
- `Avg_Pages_Viewed`
- `Avg_Customer_Rating`
- `Avg_Delivery_Time_Days`

---

### Feature Engineering

- Log-transform skewed variables:
  - `Log_Revenue`, `Log_AOV`
- Encode categorical variables:
  - One-hot encoding for nominal variables
- Scale numerical features:
  - Standardization for models sensitive to scale (e.g., Logistic Regression)

---

### Modeling Approach

#### Baseline Model
- Logistic Regression  
  → Interpretable benchmark

#### Tree-Based Models
- Random Forest  
- Gradient Boosting (e.g., XGBoost / LightGBM)

These models:
- Capture nonlinear relationships
- Handle feature interactions
- Are robust to feature scaling

---

### Evaluation Metrics

Since this is a classification problem:

- Primary:
  - **F1-score**
  - **ROC-AUC**
- Secondary:
  - Accuracy
  - Precision / Recall

If business focus is on identifying retained customers:
- Prioritize **Recall** (avoid missing loyal customers)

---

### Validation Strategy

- Train-test split (e.g., 80/20)
- Cross-validation (e.g., 5-fold)
- Check for overfitting:
  - Compare train vs test performance

---

### Model Interpretation

- Feature importance (tree-based models)
- Coefficients (logistic regression)
- SHAP values (optional, for deeper interpretability)

---

### Expected Outcome

Based on EDA:
- Models will likely rely heavily on **frequency-related features**
- Engagement variables will contribute little
- Predictive performance should be **moderate to strong**, due to clear behavioral signal
